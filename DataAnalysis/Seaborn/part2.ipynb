```python
#bar graph 
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=sns.load_dataset('penguins')
df
```

```python
sns.barplot(x='island',y='bill_length_mm',data=df)
plt.show()
```

```python
#categorical plot hue parameter
sns.barplot(x='island',y='bill_length_mm',hue='sex',data=df)
plt.show()
```

```python
order1=['Biscoe','Dream','Torgersen']
sns.barplot(x='island',y='bill_length_mm',hue='sex',data=df,order=order1)
plt.show()
```

```python
#hue order parameter can be used to change the order of the bars in the graph. The order parameter is used to specify the order of the bars in the graph. The hue parameter is
order2=['Dream','Torgersen','Biscoe']
sns.barplot(x='island',y='bill_length_mm',hue='sex',data=df,order=order2,palette='Accent')
plt.show()
```

```python
order3=['Dream','Torgersen','Biscoe']
sns.barplot(x='island',y='bill_length_mm',data=df,order=order3,color='pink')
plt.show()
```

```python
#saturation parameter can be used to change the saturation of the bars in the graph. The saturation parameter is used to specify the saturation of the bars in the graph. The hue parameter is used to specify the hue of the bars in the graph.
sns.barplot(x='island',y='bill_length_mm',hue='sex',data=df,order=order2,palette='gist_gray',saturation=0.9)   
plt.show()
```

```python
#histogram
sns.histplot(x='bill_length_mm',data=df,hue='sex',multiple='stack',palette='Set2')
plt.show()
```

```python
sns.displot(df['flipper_length_mm'])
plt.show()
```

```python
#bins
sns.displot(df['flipper_length_mm'],bins=[170,180,190,200,210,220,230,240,250])
plt.show()
```

```python
#colour
sns.displot(df['flipper_length_mm'],bins=[170,180,190,200,210,220,230,240,250],color='yellow')
plt.show()
```

```python
#kde ->kernel density estimation
sns.displot(df['flipper_length_mm'],bins=[170,180,190,200,210,220,230,240,250],kde=True,color='green')
plt.show()
```
