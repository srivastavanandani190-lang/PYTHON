Visualization is very important for analysis of data.
Matplotlib ,Seaborn and Plotly is used in python for visualization library.
Matplotlib is 2D ploting library.
Matplotlib is a multi-platform data visualization library built on NumPy arrays and designed to work with the broader SciPy stack.
It was introduced by John Hunter in the year 2002.
One of the greatest benefits of visualization is that it allows us visual access to huge amounts of data in easily digestible visuals.
pip install matplotlib.
Data Visualisation refers to graphical representation of data.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
c='r'
plt.plot(x, y,c)

plt.title("My First Graph")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.show() # only graph will be displayed when this line is executed
```

```python
x=[5,2,7,4,1]
y=[10,5,6,4,2]
plt.plot(x,y)
plt.show()
```

```python
x=[5,2,7,4,1]
y=[10,5,6,4,2]
plt.plot(x,y,marker='o')
plt.show()
```

```python
x=[5,2,7,4,1]
y=[10,5,6,9,8]
plt.plot(x,y,marker='o')
plt.plot(y,marker='>')
plt.plot(x,marker='*')
plt.show()
```

```python
x=[1, 2, 3, 4, 5]
y=[10, 20, 30, 40, 50]
plt.bar(x,y,color=['g', 'r', 'b', 'y', 'c'])
plt.show()
```

```python
#horizontal bar graph
x=['A', 'B', 'C', 'D', 'E']
y=[10, 20, 30, 40, 50]
plt.barh(x,y,color=['g', 'r', 'b', 'y', 'c'])
plt.show()
```

```python
x=[1, 2, 3, 4, 5]
y=[10, 20, 30, 40, 50]
plt.bar(x,y,color=['g', 'r', 'b', 'y', 'c'],width=0.3)
plt.show()
```

```python
x=[1, 2, 3, 4, 5]
y=[10, 20, 30, 40, 50]
plt.barh(x,y,color=['g', 'r', 'b', 'y', 'c'],height=0.3)
plt.show()
```

```python
#Scatter plot
x=[1, 2, 3, 4, 5]
y=[10, 20, 30, 40, 50]
plt.title("Scatter Plot")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.scatter(x,y,color=['g', 'r', 'b', 'y', 'c'],s=100)
plt.show()
```

```python
#changing colours of images
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

fname=r'images.jpg'

# Open the image file
img = Image.open(fname).convert('L')  # Convert to grayscale

plt.imshow(img, cmap='gray')  # Display the image in grayscale
```

```python
frame1=r'bird.jpg'
img=Image.open(frame1).convert('L')
plt.imshow(img,cmap='gray')
plt.show()
```

```python
frame1=r'bird.jpg'
img=Image.open(frame1).convert('L')
plt.imshow(img,cmap='inferno')
plt.show()
```

```python
frame1=r'bird.jpg'
img=Image.open(frame1).convert('RGB')
plt.imshow(img)
plt.show()
```

```python
#pie chart
labels = ['Coffee', 'Bread', 'Cheese', 'Dairy']
sizes = [15, 30, 45, 10]
c=['gray', 'yellowgreen', 'orange', 'blue']
plt.pie(sizes, labels=labels,colors=c)
plt.legend()
plt.show()
```
