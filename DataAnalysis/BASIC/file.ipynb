```python
#CSV ->COMMA Separated Values
csvfile=open("data.csv","w")
csvfile.write("Name,Age,City\n")
csvfile.write("Alice,25,New York\n")
csvfile.write("Bob,30,Los Angeles\n")
csvfile.write("Charlie,35,Chicago\n")
#csvfile.close()
csvfile=open("data.csv","r")
print(csvfile.read())
```

```python
#JSON -> JavaScript Object Notation
jsonfile=open("data.json","w")
jsonfile.write('{"name": "Alice", "age": 25, "city": "New York"}')
jsonfile=open("data.json","r")
print(jsonfile.read())
jsonfile=open("data.json","a")
jsonfile.write(', "country": "USA"}')
jsonfile=open("data.json","r")
print(jsonfile.read())
```
