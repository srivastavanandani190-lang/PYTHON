Functions

```python
# functions are block of code that execute certain tasks.
def greet(name="Nandani"): 
   print(f"Hello {name}")
greet("Honey")
greet()

def add(a, b):
   return a + b
print(add(5, 3))

def para(*num): #any number of arguments can be passed to the function using *args
   print(sum(num))
para(1, 2, 3, 4, 5)
para(10, 20, 30)

def kwargs1(**kwargs):#any number of keyword arguments can be passed to the function using **kwargs
   for key, value in kwargs.items():
      print(f"{key}: {value}")

kwargs1(name="Alice", age=30, city="New York")
```

```python
#Scopes
x="Global"

def myfunc():
   x="Local"
   y="locals"
   print("Inside function:", x)
myfunc()
print("Outside function:", x)
#print("Outside function:", y) #this will give an error because y is local to the function and cannot be accessed outside of it.
```

```python
#Lambda function
square = lambda x: x**2
print(square(5))
# we can't use * or ** in lambda function, it can only take one expression.

list1 = [1, 2, 3, 4, 5]
squared_list = list(map(lambda x: x**2, list1))
print(squared_list)
even_numbers = list(filter(lambda x: x % 2 == 0, list1))
print(even_numbers)
```

```python
#Modules

import math
print(math.sqrt(16))
print(math.factorial(5))
```

```python
import importlib 
#? Because Python doesn't automatically reload an imported module every time you modify its source file.
# Imports Python's module-management library.
# This concept is called Module Caching and Reloading.
import add
importlib.reload(add)

print(add.addition(5, 10))
print(add.subtraction(10, 5))
print(add.multiplication(5, 10))
print(add.division(10, 5))
```
