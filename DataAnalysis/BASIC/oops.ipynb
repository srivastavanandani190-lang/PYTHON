```python
# Classes and objects
class Car:
    def __init__(self, make, model, year):
#constructor method to initialize the attributes of the class
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        return f"{self.year} {self.make} {self.model}"

car1 = Car("BMW", "m4", 2020)
print(car1.display_info())
```
