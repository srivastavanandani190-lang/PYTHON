Data types


```python
#Numbers in python
a=2
b=3.56
c=2+3j
print(type(a))
print(type(b))
print(type(c))

#Boolean values in python
x=True
y=not x
print(type(x))
print(y)
print(4<5)

#Strings in python
name="srivani"
print(type(name))
print(name[0])  #indexing
print(name[0:4]) #slicing
print(name[-1:-5]) #negative indexing
print(name[::-1]) #reverse string
print(name.upper()) #uppercase
print(name.lower()) #lowercase
print(name.replace("s","S")) #replace just view not change the original string
print(name.split("i")) #split

#None in python
z=None
print(type(z))
```

```python
#Type conversion in python
x=5
y=2.5
print(x+y) #int+float=float implicit type conversion
print(int(y)) #float to int explicit type conversion
print(float(x)) #int to float

d="10"
e=int(d) #string to int
print(e)
print(type(e))
```

Operators


```python
#Arithmetic operators
a=81
b=3
print(a+b) #addition
print(a-b) #subtraction
print(a*b) #multiplication
print(a/b) #division
print(a//b) #floor division
print(a%b) #modulus
print(a**b) #exponentiation
print(a**0.5) #square root
print(a**(1/3)) #cube root
```

```python
#Assignment operators
x=5
x+=3 #x=x+3
print(x)
x-=2 #x=x-2
print(x)
x*=4 #x=x*4
print(x)
x/=2 #x=x/2
print(x)

```

```python
#comparison operators
x=5
print(x==5) #equal to
print(x!=5) #not equal to
print(x>5) #greater than
print(x<5) #less than
print(x>=5) #greater than or equal to
```

```python
#Logical operators
a=True
b=False
print(a and b) #logical and
print(a or b) #logical or
print(not a) #logical not
```

```python
#identity operators
x=5
y=5
z=10
print(x is y) #is operator
print(x is not y) #is not operator
print(x is z) #is operator
print(x is not z) #is not operator
```

```python
#membership operators
fruits=["apple","banana","cherry"]
print("apple" in fruits) #in operator
print("grapes" not in fruits) #not in operator
print("banana" not in fruits) #not in operator
```
