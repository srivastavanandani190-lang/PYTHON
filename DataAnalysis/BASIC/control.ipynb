```python
#conditional statements
age=int(input("enter your age:"))
if age>18:
   print("not kid")
elif age==18:
   print("just got adult")
else:
   print("kid")
```

```python
#Loops
list1=[1,2,3,4,5]
for i in list1:
    print(i)
j=1
while j<10:
    print(j)
    j+=1
r=range(10*2)
for i in r:
    if i==10:
        #break
        continue
        #pass #acts as a placeholder something to be implemented later
    print(i)
```
