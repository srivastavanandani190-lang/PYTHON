Data Structure


```python
# DS used to store data in organised form
#list is a collection of items in a particular order and mutable
list1=[21,22,"hello"]
print(list1)
print(list1[-1]) 
print(list1[0:2])  
print(type(list1))
list1[0]=100
list1.append(200)  
list1.insert(1,300)  
list1.remove(22) 
list1.pop()  
list1.pop(2) 
del list1[0] 
print(len(list1)) 
list1.clear() 
print(list1)
```

```python
#Tuple is a collection of items in a particular order and immutable
tuple1=(21,22,"hello")
print(tuple1)
print(len(tuple1))
#no append , pop any change is not possible in tuple
tuple2=(1,2,3,4,5)
print(tuple1+tuple2)
```

```python
#set is a collection of unordered items and mutable ,no indexing is possible, no duplicates are allowed
set1={1,2,3,4,5,"honey"}
print(set1)
set1.add(6)
set1.remove(3)#removes the element 3 if it exists, throws error if it doesn't exist
set1.update({7,8,9})
print(set1)
set1.discard(2)# removes the element 2 if it exists, does nothing if it doesn't
x=set1.pop()#removes a random element from the set
print(set1)
print(x)

set2={1,2,3,4,5}
set3={4,5,6,7,8}
print(set2.union(set3))#returns a new set with all elements from both sets
print(set2.intersection(set3))#returns a new set with elements common to both sets
print(set2.difference(set3))#returns a new set with elements in set2 but not in set3
```

```python
#dictionary is a collection of key-value pairs, unordered and mutable
dict1={"name":"John","age":30,"city":"New York"}
print(len(dict1))
print(dict1["name"])#accessing value using key
dict1["age"]=31#updating value
dict1["country"]="USA"#adding new key-value pair
dict1.pop("city")#removes the key-value pair with key "city"
print(dict1)
dict1.popitem()#removes the last inserted key-value pair
print(dict1)
```
